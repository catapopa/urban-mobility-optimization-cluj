```python
import osmnx as ox
import networkx as nx
import matplotlib.pyplot as plt

# Load graph
G = ox.load_graphml("../data/processed/cluj_osm.graphml")

# Show summary
print(ox.basic_stats(G))

# Calculate and plot betweenness centrality (just for top nodes for speed)
bc = nx.betweenness_centrality(G, k=100, normalized=True, endpoints=True)

# Add as node attribute
nx.set_node_attributes(G, bc, 'betweenness')

# Get top nodes
top_nodes = sorted(bc.items(), key=lambda x: x[1], reverse=True)[:10]
print("Top 10 nodes by betweenness centrality:")
for node, value in top_nodes:
    print(f"Node {node}: {value:.4f}")

# Plot with centrality color
fig, ax = ox.plot_graph(G, node_color=list(bc.values()), node_size=10, edge_color='gray', bgcolor='white')
```
